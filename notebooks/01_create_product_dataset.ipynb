```python
import requests
import os
import time
from pypdf import PdfReader

# Create folder for downloaded PDFs
os.makedirs("../tds_pdfs", exist_ok=True)

# List to store TDS information
tds_data = []

for index, row in df.iterrows():

    product_id = row["Product_ID"]
    product_name = row["Product_Name"]
    tds_url = row["TDS_Link"]

    print(f"Processing {product_id}: {product_name}")

    # Skip products without a TDS link
    if pd.isna(tds_url):
        tds_data.append({
            "Product_ID": product_id,
            "Product_Name": product_name,
            "TDS_Link": None,
            "TDS_Text": None,
            "Status": "TDS Link Missing"
        })
        continue

    try:
        # Download PDF
        response = requests.get(tds_url, timeout=30)
        response.raise_for_status()

        # Save PDF
        pdf_path = f"../tds_pdfs/{product_id}.pdf"

        with open(pdf_path, "wb") as f:
            f.write(response.content)

        # Extract PDF text
        reader = PdfReader(pdf_path)

        text = ""

        for page in reader.pages:
            page_text = page.extract_text()

            if page_text:
                text += page_text + "\n"

        # Store result
        tds_data.append({
            "Product_ID": product_id,
            "Product_Name": product_name,
            "TDS_Link": tds_url,
            "TDS_Text": text.strip(),
            "Status": "Success"
        })

        print("   ✓ Downloaded and extracted")

    except Exception as e:

        tds_data.append({
            "Product_ID": product_id,
            "Product_Name": product_name,
            "TDS_Link": tds_url,
            "TDS_Text": None,
            "Status": f"Error: {str(e)}"
        })

        print("   ✗ Error:", e)

    # Small delay between downloads
    time.sleep(0.5)

print("Finished processing all TDS files.")
```

```python
tds_df = pd.DataFrame(tds_data)

print(tds_df.shape)
print(tds_df["Status"].value_counts())
```

```python
import re

def clean_text_for_excel(text):
    if pd.isna(text):
        return text

    text = str(text)

    # Remove characters that Excel does not allow
    text = re.sub(r'[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]', '', text)

    return text
```

```python
tds_df["TDS_Text"] = tds_df["TDS_Text"].apply(clean_text_for_excel)
```

```python
import os

os.makedirs("../output", exist_ok=True)

tds_output = "../output/tds_dataset.xlsx"

tds_df.to_excel(tds_output, index=False)

print("TDS dataset saved successfully!")
print("Location:", tds_output)
```

```python
import requests
import os
import time
import pandas as pd
from pypdf import PdfReader

# Create folder
os.makedirs("../msds_pdfs", exist_ok=True)

msds_data = []

for index, row in df.iterrows():

    product_id = row["Product_ID"]
    product_name = row["Product_Name"]
    msds_url = row["MSDS_Link"]

    print(f"Processing {product_id}: {product_name}")

    # If MSDS link is missing
    if pd.isna(msds_url) or str(msds_url).strip() == "":
        msds_data.append({
            "Product_ID": product_id,
            "Product_Name": product_name,
            "MSDS_Link": None,
            "MSDS_Text": None,
            "Status": "MSDS Link Missing"
        })

        print("   - MSDS link missing")
        continue

    try:
        # Download PDF
        response = requests.get(msds_url, timeout=30)
        response.raise_for_status()

        # Save PDF
        pdf_path = f"../msds_pdfs/{product_id}.pdf"

        with open(pdf_path, "wb") as f:
            f.write(response.content)

        # Extract text
        reader = PdfReader(pdf_path)

        text = ""

        for page in reader.pages:
            page_text = page.extract_text()

            if page_text:
                text += page_text + "\n"

        msds_data.append({
            "Product_ID": product_id,
            "Product_Name": product_name,
            "MSDS_Link": msds_url,
            "MSDS_Text": text.strip(),
            "Status": "Success"
        })

        print("   ✓ Downloaded and extracted")

    except Exception as e:

        msds_data.append({
            "Product_ID": product_id,
            "Product_Name": product_name,
            "MSDS_Link": msds_url,
            "MSDS_Text": None,
            "Status": f"Error: {str(e)}"
        })

        print("   ✗ Error:", e)

    # Small delay so we don't send requests too quickly
    time.sleep(0.5)

print("\nFinished processing all MSDS files.")
```

```python
msds_df = pd.DataFrame(msds_data)

print("Dataset shape:", msds_df.shape)

print("\nStatus:")
print(msds_df["Status"].value_counts())
```

```python
import re

def clean_text_for_excel(text):

    if pd.isna(text):
        return text

    text = str(text)

    # Remove characters that Excel cannot store
    text = re.sub(
        r'[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]',
        '',
        text
    )

    return text


msds_df["MSDS_Text"] = msds_df["MSDS_Text"].apply(
    clean_text_for_excel
)
```

```python
import pandas as pd

# Load the three datasets
products_df = pd.read_excel("../output/products_dataset.xlsx")
tds_df = pd.read_excel("../output/tds_dataset.xlsx")
msds_df = pd.read_excel("../output/msds_dataset.xlsx")

# Keep only TDS information that is not already in products_df
tds_master = tds_df[
    ["Product_ID", "TDS_Text", "Status"]
].copy()

tds_master = tds_master.rename(
    columns={"Status": "TDS_Status"}
)

# Keep only MSDS information that is not already in products_df
msds_master = msds_df[
    ["Product_ID", "MSDS_Text", "Status"]
].copy()

msds_master = msds_master.rename(
    columns={"Status": "MSDS_Status"}
)

# Merge TDS data
master_df = products_df.merge(
    tds_master,
    on="Product_ID",
    how="left"
)

# Merge MSDS data
master_df = master_df.merge(
    msds_master,
    on="Product_ID",
    how="left"
)

# Arrange columns
master_df = master_df[
    [
        "Product_ID",
        "Category",
        "Product_Name",
        "TDS_Link",
        "TDS_Text",
        "TDS_Status",
        "MSDS_Link",
        "MSDS_Text",
        "MSDS_Status"
    ]
]

# Check result
print("Master dataset shape:", master_df.shape)
print("\nColumns:")
print(master_df.columns.tolist())

# Save
master_df.to_excel(
    "../output/hindcon_master_dataset.xlsx",
    index=False
)

print("\nMaster dataset created successfully!")
```
